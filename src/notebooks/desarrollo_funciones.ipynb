```python
# Histobrama por categorías, todo en una
def histograma_kde_por_categoria(df, var_cuantitativa, var_cualitativa, bins=20, estadistico = "count"):
    plt.figure(figsize=(10, 6))

    sns.histplot(
        data=df,
        x=var_cuantitativa,
        hue=var_cualitativa,
        bins=bins,
        kde=True,
        stat=estadistico,   # "density" para valores relativos
        common_norm=False,  # si se usa density, normaliza respecto del dataset completo
        alpha=0.4
    )

    plt.title(f"Distribución de {var_cuantitativa} según {var_cualitativa}")
    plt.xlabel(var_cuantitativa)
    plt.ylabel("Densidad")
    plt.grid(alpha=0.3)

    plt.show()
```

```python
# Histograma por grupos de categorías, separados
def histograma_kde_por_categoria(df, var_cuantitativa, var_cualitativa, bins=20, max_por_grafico = 3,  estadistico = "count"):
    categorias = df[var_cualitativa].unique()
    grupos = [categorias[i:i + max_categorias_por_grafico]
        for i in range(0, len(categorias), max_categorias_por_grafico)]
    
    for grupo in grupos:
        plt.figure(figsize=(10, 6))

        sns.histplot(
            data=df,
            x=var_cuantitativa,
            hue=var_cualitativa,
            hue_order = grupo,  # limitar a las variables en el grupo
            bins=bins,
            kde=True,
            stat=estadistico,   # "density" para valores relativos
            common_norm=False,  # si se usa density, normaliza respecto del dataset completo
            alpha=0.4
        )

        plt.title(f"Distribución de {var_cuantitativa} según {var_cualitativa}")
        plt.xlabel(var_cuantitativa)
        plt.ylabel("Densidad")
        plt.grid(alpha=0.3)

        plt.show()
    return
```
