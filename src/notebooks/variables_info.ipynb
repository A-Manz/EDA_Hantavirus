Para este análisis de los casos y distribución del Hantavirus, se utilizarán variables de la tabla ```clinical```
### Variable dependiente
```y: length of stay days``` (int)
### Variables independientes
```x_1: country``` - cualitativa, nombre del país \
```x_2: syndrome``` - cualitativa binaria (HPS o HFRS) \
```x_3: age``` - cuantitativa discreta (int) \
```x_4: sex``` - cualitativa binaria (Male, Female) \
```x_5: severity``` - cualitativa, 4 valores: Mild/Moderate/Severe/Critical \
```x_6: outcome``` - cualitativa binaria (Recovered/Deceased)

### Para la regresión lineal:
Las variables cualitativas, exceptuando el país, serán traducidas a enteros en orden creciente:
- Las variables binarias serán transformadas a variables 0/1
- La variable ```severity``` constará con cuatro valores: 0, 1, 2 y 3; estos corresponden, en orden, a los valores Mild, Moderate, Severe y Critical
